## SKILLS
* 공식 문서
    * https://code.claude.com/docs/en/skills 
* Skill 은 instruction (지시 방법)
    * 즉, 프롬프트
    * 자주 쓰는 기능들을 재사용성을 높인 프롬프트 셋
    * SKill은 구조가 있음

```
my-skill/
├── SKILL.md           # Main instructions (required)
├── template.md        # Template for Claude to fill in
├── examples/
│   └── sample.md      # Example output showing expected format
└── scripts/
    └── validate.sh    # Script Claude can execute (python, js ...)
```

* 복잡한 기능을 구현하는게 아니라면 SKILL.md 만으로 충분함

### SKILL 작성 팁
* 공식 문석의 내용 복사
* `<conetext> [paste] </context> 지시사항`
* '답변을 생성하기 위해 필요한 정보가 있다면 나에게 물어봐라' 라는 프롬프트 기법을 활용
* SKILL을 개선하기 위해 아래의 문서 내용을 context 로 주입
    * https://platform.claude.com/docs/en/agents-and-tools/agent-skills/best-practices
